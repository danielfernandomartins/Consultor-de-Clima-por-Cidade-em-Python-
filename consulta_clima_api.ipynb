```python
# Importa a biblioteca requests.
# Ela permite fazer requisições para APIs na internet.
import requests


# Cria uma função chamada buscar_cidade.
# Essa função procura uma cidade e retorna suas coordenadas.
def buscar_cidade(nome_cidade):

    # Define o endereço da API de geolocalização.
    url = "https://geocoding-api.open-meteo.com/v1/search"

    # Cria os parâmetros que serão enviados para a API.
    parametros = {

        # Envia o nome da cidade digitada pelo usuário.
        "name": nome_cidade,

        # Solicita apenas um resultado.
        "count": 1,

        # Solicita os dados em português.
        "language": "pt",

        # Define o formato da resposta como JSON.
        "format": "json"
    }

    # Faz uma requisição GET para a API.
    resposta = requests.get(url, params=parametros)

    # Verifica se a requisição funcionou corretamente.
    if resposta.status_code == 200:

        # Converte a resposta da API para um dicionário Python.
        dados = resposta.json()

        # Verifica se a API encontrou algum resultado.
        if "results" in dados:

            # Pega o primeiro resultado encontrado.
            cidade = dados["results"][0]

            # Retorna os dados da cidade.
            return cidade

        # Caso nenhuma cidade seja encontrada.
        else:

            # Retorna None.
            return None

    # Caso aconteça algum erro na requisição.
    else:

        # Retorna None.
        return None


# Cria uma função chamada buscar_clima.
# Ela consulta o clima usando latitude e longitude.
def buscar_clima(latitude, longitude):

    # Define o endereço da API de previsão do tempo.
    url = "https://api.open-meteo.com/v1/forecast"

    # Cria os parâmetros da consulta.
    parametros = {

        # Envia a latitude da cidade.
        "latitude": latitude,

        # Envia a longitude da cidade.
        "longitude": longitude,

        # Solicita informações do clima atual.
        "current": (
            "temperature_2m,"
            "relative_humidity_2m,"
            "apparent_temperature,"
            "wind_speed_10m"
        ),

        # Faz a API usar automaticamente o horário da cidade.
        "timezone": "auto"
    }

    # Faz a requisição GET para a API.
    resposta = requests.get(url, params=parametros)

    # Verifica se a requisição foi bem-sucedida.
    if resposta.status_code == 200:

        # Converte a resposta para JSON.
        dados = resposta.json()

        # Retorna os dados recebidos.
        return dados

    # Caso exista algum erro.
    else:

        # Retorna None.
        return None


# Exibe o título do programa.
print("====================================")

# Exibe o nome do programa.
print("      CONSULTA DE CLIMA")

# Exibe outra linha decorativa.
print("====================================")


# Solicita ao usuário o nome da cidade.
nome_cidade = input("\nDigite o nome da cidade: ")


# Chama a função que pesquisa a cidade.
cidade = buscar_cidade(nome_cidade)


# Verifica se a cidade foi encontrada.
if cidade is not None:

    # Guarda o nome encontrado pela API.
    nome = cidade["name"]

    # Guarda o estado.
    # O comando get evita erro caso essa informação não exista.
    estado = cidade.get("admin1", "Não informado")

    # Guarda o nome do país.
    pais = cidade.get("country", "Não informado")

    # Guarda a latitude.
    latitude = cidade["latitude"]

    # Guarda a longitude.
    longitude = cidade["longitude"]


    # Mostra a cidade encontrada.
    print("\nCidade encontrada:")

    # Exibe o nome.
    print(f"Cidade: {nome}")

    # Exibe o estado.
    print(f"Estado: {estado}")

    # Exibe o país.
    print(f"País: {pais}")


    # Chama a função que consulta o clima.
    clima = buscar_clima(latitude, longitude)


    # Verifica se os dados do clima foram encontrados.
    if clima is not None:

        # Guarda os dados do clima atual.
        atual = clima["current"]

        # Guarda a temperatura.
        temperatura = atual["temperature_2m"]

        # Guarda a sensação térmica.
        sensacao = atual["apparent_temperature"]

        # Guarda a umidade.
        umidade = atual["relative_humidity_2m"]

        # Guarda a velocidade do vento.
        vento = atual["wind_speed_10m"]


        # Mostra um título.
        print("\n--- CLIMA ATUAL ---")

        # Mostra a temperatura.
        print(f"Temperatura: {temperatura} °C")

        # Mostra a sensação térmica.
        print(f"Sensação térmica: {sensacao} °C")

        # Mostra a umidade.
        print(f"Umidade: {umidade}%")

        # Mostra a velocidade do vento.
        print(f"Velocidade do vento: {vento} km/h")


    # Caso aconteça erro na consulta do clima.
    else:

        # Mostra mensagem de erro.
        print("\nNão foi possível consultar o clima.")


# Caso a cidade não seja encontrada.
else:

    # Mostra mensagem para o usuário.
    print("\nCidade não encontrada.")
```
